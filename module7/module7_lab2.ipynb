```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

df = pd.read_csv("ratings.tsv", sep="\t")

print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
X = df[["ambience", "food", "service"]]
y = df["rating"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

print("Test R² score:", model.score(X_test, y_test))
print("Intercept:", model.intercept_)
```

```python
coefficients = pd.DataFrame({
    "Factor": X.columns,
    "Coefficient": model.coef_
})

coefficients = coefficients.sort_values(
    "Coefficient", ascending=False
).reset_index(drop=True)

coefficients
```

So from my results, I found that food had the highest coefficient at around 0.96, followed by service at 0.20 and ambience at 0.07. So, if I were to run a restaurant that was struggling with its overall rating, first things first, I would focus on improving the food as it would have the strongest relationship with the overall rating. That being said, that doesn't necessarily mean improving the food alone would guarantee better ratings.
