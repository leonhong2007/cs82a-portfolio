```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")

print(df.shape)
display(df.head(10))
print(df.dtypes)
```

Initial Observations:
-Some values in the `price` column are missing.
-The `zip` column is stored as a number, which can remove leading zeros.
-The `date` column is stored as text instead of a datetime type.

```python
df.isna().sum()
```

```python
median_price = df["price"].median()

print("Median price:", median_price)

df["price"] = df["price"].fillna(median_price)

df.isna().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("New shape:", df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)

df[["zip"]].head(10)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")

print(df.dtypes)
```

```python
print("Shape:", df.shape)
print("\nMissing values:")
print(df.isna().sum())
print("\nZIP lengths:")
print(df["zip"].str.len().value_counts())
```

CLeaning LOG

I started with 300 rows and 6 columns.

-There were 12 missing prices, so I filled them in with the median price of $37.53. I used the median because really high or low prices would affect it less.

-I found 8 duplicate rows and removed them, which left 292 rows.

-I fixed the ZIP codes by changing them to text then adding back the leading zeros so that they were all 5 digits.

-I changed the date column into datetime format so all the dates were stored the same way.

-I found 3 rows with negative quantities. I kept them because they could just be returns and not actual mistakes.

If I were to clean the data incorrectly, it could mess up the sales numbers and make the business come to an incorrect conclusion.
