```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_clean.csv")

print(df.shape)
df.head()
```

```python
print("Price Mean:", df["price"].mean())
print("Price Median:", df["price"].median())
print("Price Standard Deviation:", df["price"].std())

print("Quantity Mean:", df["qty"].mean())
print("Quantity Median:", df["qty"].median())
print("Quantity Standard Deviation:", df["qty"].std())
```

The price I would use is the median of 37.53 due to the fact that the average price of $55.52 is a lot higher. Which shows that a few expensive items are putting the average up, so the median gives representation of what the average item costs.

```python
df["price"].plot(kind="hist", bins=20)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

The histogram is right-skewed because most items have lower prices, while only a few items have really high prices. Which also explains why the mean price is higher than the median.

```python
df["revenue"] = df["price"] * df["qty"]

print(df.shape)
df.head()
```

```python
corr_matrix = df.corr(numeric_only=True)

print(corr_matrix)
```

```python
df.plot.scatter(x="price", y="revenue")
plt.title("Price vs Revenue")
plt.show()
```

The strongest correlation is between price and revenue (r = 0.818), showing a strong positive relationship

```python
df.plot.scatter(x="qty", y="revenue")
plt.title("Quantity vs Revenue")
plt.show()
```

```python
print("Price vs Revenue:", df["price"].corr(df["revenue"]))
print("Quantity vs Revenue:", df["qty"].corr(df["revenue"]))
```

What the data shoes is a strong positive correlation between price and revenue (r = 0.818), which means that higher-priced items generally bring in more revenue, but due to the fact that revenue also depends on amount sold, we shouldn't assume raising prices will automatically increase revenue.
