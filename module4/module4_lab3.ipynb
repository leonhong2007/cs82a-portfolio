```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("truck.csv")

print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
r = df["X"].corr(df["Y"])

print("Pearson correlation (r):", r)
```

```python
df.plot.scatter(x="X", y="Y")

plt.title("Truck List Prices and Sale Prices Are Strongly Related")
plt.xlabel("List Price (thousands of dollars)")
plt.ylabel("Best Sale Price (thousands of dollars)")
plt.show()
```

```python
print("Minimum list price:", df["X"].min())
print("Maximum list price:", df["X"].max())

print(df[["X", "Y"]].describe())
```

```python
import numpy as np

slope, intercept = np.polyfit(df["X"], df["Y"], 1)

list_price = 18
predicted_price = slope * list_price + intercept

print("List Price: $", list_price * 1000)
print("Predicted Sale Price: $", round(predicted_price * 1000, 2))
```

Based off the data, I would think that 18,000 USD truck would sell for around 15,755USD. I think it will sell at this price because the prices on the graph follow almost an exactly perfectly straight line, showcasing a really strong relationship. However, this doesn't mean the prediction will always be one to one accurate, especially for trucks priced outside the range of 12,400 USD to 22,400 USD that we looked at.
