```python
import pandas as pd

df = pd.read_csv("allbacks.csv")

print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
print("Cover counts:")
print(df["cover"].value_counts())

print("\nSorted cover types:")
print(df["cover"].sort_values().tolist())

print("\nMedian cover type:")
print(df["cover"].sort_values().iloc[len(df) // 2])
```

```python
print("Overall Mean Weight:", df["weight"].mean())

print("\nMean Weight by Cover Type:")
print(df.groupby("cover")["weight"].mean())
```

Median Cover Type: The median cover type is paperback (pb). This is because there 8 paperbacks and 7 hardbacks, so when you line them up in order, paperback ends up being the middle value.
Overall Mean Weight: The weight averaged with all 15 books is around 706.67 grams. What this show is a good idea of how much each book weighs on average.
Comparison by Cover Type: When comparing the weights, hardbacks weighed about 796.43 grams on average, compared to 628.13 grams for paperbacks. So, hardbacks are about 168 grams heavier, which could increase shipping costs/complications.

By-Hand Verification: Overall Mean Weight

Hardbacks: 7 × 796.428571 ≈ 5575 grams
Paperbacks: 8 × 628.125 = 5025 grams

Total weight = 5575 + 5025 = 10600 grams

Mean = 10600 / 15 = 706.67 grams
This matches the Python result!
