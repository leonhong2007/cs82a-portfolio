```python
import sqlite3
import pandas as pd

con = sqlite3.connect("module6_lab2.db")

con.execute("DROP TABLE IF EXISTS customers")

con.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    total_spent REAL
)
""")

con.commit()

print("Customers table created successfully.")
```

```python
customers = [
    (1, "Alex Kim", "Los Angeles", 250.50),
    (2, "Jordan Lee", "San Diego", 180.25),
    (3, "Taylor Park", "Irvine", 320.00),
    (4, "Morgan Chen", "Santa Monica", 145.75),
    (5, "Sam Rivera", "Long Beach", 410.20),
    (6, "Casey Johnson", "Pasadena", 95.50),
    (7, "Jamie Brown", "Torrance", 275.00),
    (8, "Riley Davis", "Culver City", 195.80)
]

con.executemany("""
INSERT INTO customers (id, name, city, total_spent)
VALUES (?, ?, ?, ?)
""", customers)

con.commit()

print("8 fictional customers inserted successfully.")
```

```python
results = pd.read_sql("SELECT * FROM customers", con)

results
```

One difference I noticed between MySQL and SQLite was on how they handled each type of data. MySQL most commonly used VARCHAR for text, while SQLite used TEXT. For specific assignment, I used TEXT for names and cities, INTEGER for IDs, and REAL for spending amounts.
