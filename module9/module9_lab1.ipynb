```python
def pounds_to_kg(pounds):
    kilograms = pounds * 0.453592
    return kilograms
```

```python
print("100 pounds =", pounds_to_kg(100), "kg")
print("200 pounds =", pounds_to_kg(200), "kg")
print("260 pounds =", pounds_to_kg(260), "kg")
```

So this notebook uses a Python function to convert pounds into kilograms. What I did was I test three different weights using the conversion factor of 0.453592. The notebook was executed using Google Colab's cloud runtime.
