```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("custdata.tsv", sep="\t")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

df.head()
```

```python
columns = ["age", "income", "num.vehicles"]

before_corr = df[columns].corr()

print("BEFORE CLEANING:")
print(before_corr)
```

```python
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())

print("Minimum income:", df["income"].min())
print("Maximum income:", df["income"].max())

print("Negative incomes:", (df["income"] < 0).sum())
print("Negative ages:", (df["age"] < 0).sum())
print("Ages over 120:", (df["age"] > 120).sum())

print("Missing values:")
print(df[columns].isna().sum())
```

```python
clean_df = df[
    df["age"].between(0, 120) &
    (df["income"] >= 0) &
    df["num.vehicles"].notna()
].copy()

print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("Total rows removed:", len(df) - len(clean_df))

print("\nAFTER CLEANING:")
after_corr = clean_df[columns].corr()
print(after_corr)
```

```python
print("BEFORE CLEANING:")
print(before_corr.round(3))

print("\nAFTER CLEANING:")
print(after_corr.round(3))

print("\nRows removed by reason:")
print("Age over 120:", (df["age"] > 120).sum())
print("Negative income:", (df["income"] < 0).sum())
print("Missing vehicles:", df["num.vehicles"].isna().sum())
```

When cleaning the data, I removed 64 rows from the original 1,000 because some of the information was either missing or didn't make sense. There were 7 rows with ages over 120, 1 with a negative income, and 56 with missing vehicle information. The reason I removed these is because they could've messed with the correlation results, leaving me with 936 rows to work with.

After cleaning the data, what I found was that age and income barely had any correlation (r = 0.007). On the other hand the relationship between age and the # of vehicles was also quite weak, with a slight negative correlation (r = -0.072). Out of the three, income and the number of vehicles had the strongest correlation (r = 0.134), but even then, the relationship was still pretty weak.
