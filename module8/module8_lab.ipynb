**First AI Prompt:**

I'm working on an exploratory data analysis assignment in Google Colab using a penguins dataset with 344 rows and 8 columns.

My columns and data types are:
- species: object
- island: object
- bill_length_mm: float64
- bill_depth_mm: float64
- flipper_length_mm: float64
- body_mass_g: float64

Can you write Python code using pandas and seaborn to check for missing values, compare average flipper lengths across the three penguin species, and make a labeled bar chart? Please include comments explaining the code and account for missing values.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Check missing values
print("Missing values by column:")
print(df.isna().sum())

# Calculate average flipper length for each species
species_means = df.groupby("species")["flipper_length_mm"].mean()

print("\nAverage flipper length by species:")
print(species_means)

# Create a bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=species_means.index,
    y=species_means.values
)

plt.title("Average Flipper Length by Penguin Species")
plt.xlabel("Penguin Species")
plt.ylabel("Average Flipper Length (mm)")
plt.ylim(bottom=0)

plt.tight_layout()
plt.show()
```

### Follow-Up Prompt

Can you create a box plot comparing the flipper lengths of the three penguin species? I want to see how much the measurements vary within each species instead of just looking at the averages.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="species",
    y="flipper_length_mm"
)

plt.title("Flipper Length Distribution by Penguin Species")
plt.xlabel("Penguin Species")
plt.ylabel("Flipper Length (mm)")

plt.tight_layout()
plt.show()
```

So I kept the box plot because what it showed was how much flipper lengths varied between from penguin species to penguin species, which wasn't something I antipicated from the original bar chart.
