Cutoffs: 32°F or below = freezing, 33–50°F = cold, 51–65°F = mild, 66–80°F = warm, and above 80°F = hot.

```python
temps = [25, 40, 55, 65, 72, 85, 32, 50, 80]

for temp in temps:
    if temp <= 32:
        label = "freezing"
    elif temp <= 50:
        label = "cold"
    elif temp <= 65:
        label = "mild"
    elif temp <= 80:
        label = "warm"
    else:
        label = "hot"

    print(temp, "-", label)
```
