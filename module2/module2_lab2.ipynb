Problem 4.2 - Multiplication!

```python
def multiply_for(a, b):
    result = 0

    for i in range(b):
        result = result + a

    return result
```

```python
print(multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    result = 0
    count = 0

    while count < b:
        result = result + a
        count = count + 1

    return result
```

```python
print(multiply_while(7, 6))
```

```python
print("For loop:", multiply_for(7, 6))
print("While loop:", multiply_while(7, 6))
print("Real product:", 7 * 6)
```
